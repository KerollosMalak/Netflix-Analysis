```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='type', palette='Set2')
plt.title('Number of Movies vs TV Shows on Netflix')
plt.xlabel('Content Type')
plt.ylabel('Count')
# plt.grid(axis='y')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
df['year_added'].value_counts().sort_index().plot(kind='bar', color='skyblue')
plt.title('Content Added per Year on Netflix')
plt.xlabel('Year')
plt.ylabel('Number of Titles')
# plt.grid(axis='y')
plt.show()
```

```python
top_countries = df['country'].value_counts().head(10)

plt.figure(figsize=(10, 6))
top_countries.plot(kind='barh', color='salmon')
plt.title('Top 10 Countries Producing Netflix Content')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.gca().invert_yaxis()
# plt.grid(axis='x')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, y='rating', order=df['rating'].value_counts().index[:10], palette='muted')
plt.title('Top Netflix Ratings')
plt.xlabel('Count')
plt.ylabel('Rating')
# plt.grid(axis='x')
plt.show()
```

```python
from collections import Counter

genre_counter = Counter()
for genres in df['genres'].dropna():
    genre_counter.update([genre.strip() for genre in genres])

top_genres = pd.DataFrame(genre_counter.most_common(10), columns=['Genre', 'Count'])

plt.figure(figsize=(10, 6))
sns.barplot(data=top_genres, x='Count', y='Genre', palette='pastel')
plt.title('Top 10 Genres on Netflix')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
# plt.grid(axis='x')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='dayofweek_added', order=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'], palette='cool')
plt.title('Content Added by Day of the Week')
plt.xlabel('Day')
plt.ylabel('Count')
# plt.grid(axis='y')
plt.show()
```
